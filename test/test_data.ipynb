```python
import pickle
import numpy as np

# 读取pkl文件的函数
def load_pkl_file(file_path):
    try:
        with open(file_path, 'rb') as file:  # 以二进制方式打开文件
            data = pickle.load(file)  # 反序列化数据
            print("文件加载成功！")
            return data
    except Exception as e:
        print(f"加载文件时出错: {e}")
        return None
```
